```python
titanic_np = titanic_df[['Age','Fare','Survived']].to_numpy()
print(type(titanic_np)) 
# using to_numpy() function we can directly convert Pandas Series to numpy
```

```python
print(titanic_df.shape)
```

```python
titanic_df = titanic_df[['Age', 'Fare', 'Survived']].dropna()
print(titanic_df.shape)
print(titanic_df.head(10))
#Dropna drops values which has empty cell across these columns
```

```python
titanic_df['Z_norm_age'] = (titanic_df['Age'] - titanic_df['Age'].mean()) / titanic_df['Age'].std()
titanic_df['Z_norm_fare'] = (titanic_df['Fare'] - titanic_df['Fare'].mean()) / titanic_df['Fare'].std()
print(titanic_df[['Age', 'Z_norm_age', 'Fare', 'Z_norm_fare']].head(10))
#Z=X-mean/Std deviation
```

```python
mean_values = titanic_df.groupby('Survived')[['Age', 'Fare']].mean()
print(mean_values)
#groupby can group same category of data
```

```python
fare_mean = titanic_df['Fare'].mean()
titanic_df['Fare_Class'] = np.where(titanic_df['Fare'] < fare_mean, 'Low', 'High')
print(titanic_df[['Fare', 'Fare_Class']].head(10))
#np.where can be used as (condition, value_if_true  ,value_if_false)
```

```python
iris_df['petal_ratio']=iris_df['petal.length']/iris_df['petal.width']
iris_df.head(10)
#adding of new column which has length/width
```

```python
iris_df['average_petal_ratio_per_species']=iris_df.groupby(by='variety')['petal_ratio'].transform('mean')
print(iris_df.head(10))
print(iris_df.tail(10))
```

```python
print(iris_df.groupby(by='variety')['sepal.length'].std())
print(iris_df.groupby(by='variety')['sepal.length'].std().max())
```
